## 2. The Question

**Question:** Which content pages should a reviewer look at first, given
limited review capacity?

**Unit of analysis:** one content page (`content_id`), evaluated at a single
point in time.

**Output:** a ranked list of pages with a priority score and a reason code
explaining why each page was flagged.

**The action:** a human reviewer opens each top-ranked page and decides on
one of: refresh, expand, protect, prune, or monitor.

**Cost of a wrong call:**
- *False positive* (page flagged but didn't need attention): low cost — a
  reviewer spends a few minutes on a page that turns out fine.
- *False negative* (a genuinely declining, high-traffic page is missed):
  higher cost — the page keeps losing visibility, and the loss compounds
  the longer it goes unnoticed.

**Why data/ML can help at all:** a fixed rule applies the same weights to
every page regardless of context. A learned model can pick up on
*interactions* between signals — for example, "declining AND thin AND
previously high-ranking" may matter far more together than any single
signal alone — patterns that are hard to hand-write into a rule.
